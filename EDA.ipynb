```python
# Exploratory Data Analysis: Planet Zoo 2 Animals Dataset
#This project performs an exploratory data analysis (EDA) of the Planet Zoo 2 launch animal dataset.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Data Overview
##The dataset contains 130 animals and 28 variables. It includes information about animal classification, conservation status, habitat requirements, size, weight, and barrier requirements.
```

```python
animals = pd.read_csv("animals.csv")

animals.head()
```

```python
## Numerical Variables
#The numerical variables are examined using histograms and summary statistics to look at their distributions, averages, ranges, and possible outliers.
```

```python
sns.histplot(data=animals, x="size_male")
plt.title("Distribution of Male Animal Size")
plt.xlabel("Male Size")
plt.show()

sns.histplot(data=animals, x="size_female")
plt.title("Distribution of Female Animal Size")
plt.xlabel("Female Size")
plt.show()
```

```python
### Male and Female Size
#Male size ranges from 0.014 to 5.4, with an average of 1.25 and a median of 0.90. Female size ranges from 0.0155 to 5.0, with an average of 1.17 and a median of 0.88.
#Both graphs are right skewed, with most of the animals having smaller size measurements. There are only a few animals with measurements around 4 to 5, which could be possible outliers. The male and female distributions also look pretty similar overall.
```

```python
sns.histplot(data=animals, x="weight_male_kg")
plt.title("Distribution of Male Animal Weight")
plt.xlabel("Male Weight (kg)")
plt.show()

sns.histplot(data=animals, x="weight_female_kg")
plt.title("Distribution of Female Animal Weight")
plt.xlabel("Female Weight (kg)")
plt.show()
```

```python
### Male and Female Weight
#The male weights range from 0.001 kg to 1750 kg, with an average of 147.90 kg and a median of 23.50 kg. The female weights range from 0.0012 kg to 1250 kg, with an average of 116.75 kg and a median of 23.50 kg.
#Both graphs are very right skewed. Most of the animals have lower weights, but there are a few animals that weigh a lot more than the rest. This is also why the averages are much higher than the medians. These larger values could be outliers, which will be looked at more closely using the box plots.
```

```python
sns.histplot(data=animals, x="barrier_grade")
plt.title("Distribution of Barrier Grade")
plt.xlabel("Barrier Grade")
plt.show()

sns.histplot(data=animals, x="barrier_min_height_m")
plt.title("Distribution of Minimum Barrier Height")
plt.xlabel("Minimum Barrier Height (m)")
plt.show()
```

```python
### Barrier Grade and Minimum Barrier Height
#Barrier grade ranges from 1 to 4, with an average of 2.19 and a median of 2. Most animals with barrier data have a grade of 2, while grades 1 and 4 are less common.
#Minimum barrier height ranges from 0.5 m to 3.0 m, with an average of 1.61 m and a median of 1.25 m. Most of the animals require barriers around 1 to 1.5 meters, but there is also a noticeable group that requires a 3-meter barrier.
```

```python
numerical_cols = [
    "size_male",
    "size_female",
    "weight_male_kg",
    "weight_female_kg",
    "barrier_grade",
    "barrier_min_height_m"
]
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=animals[numerical_cols])

plt.title("Box Plot of Numerical Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
### Numerical Variables Box Plot
#The box plot shows all six numerical variables together. The weight variables have a much larger range than the other variables, which makes the size and barrier variables look compressed near the bottom of the graph.
#There are several noticeable outliers in both male and female weight. Male weight has the highest outlier at 1750 kg, while female weight has the highest at 1250 kg. There also appear to be some possible outliers in the size variables, but they are harder to see on this graph because of the difference in scale.
```

```python
numerical_data = animals.select_dtypes(include="number")

Q1 = numerical_data.quantile(0.25)
Q3 = numerical_data.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((numerical_data < lower_bound) |
            (numerical_data > upper_bound)).sum()

outliers
```

```python
sns.countplot(data=animals, x="size_measurement")
plt.title("Animals by Size Measurement")
plt.xlabel("Size Measurement")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=animals, x="conservation_status")
plt.title("Animals by Conservation Status")
plt.xlabel("Conservation Status")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.countplot(data=animals, x="class")
plt.title("Animals by Class")
plt.xlabel("Class")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
### Categorical Variable Observations
##The enclosure type graph shows that HabitatTerrestrial is the most common enclosure type, followed by Exhibit. For size measurements, length is the most common measurement used.
##Most of the animals are classified as Least Concern, while some of the other conservation statuses have much smaller counts. Mammalia is also the most common animal class in the dataset, followed by Actinopterygii.
```

```python
## Pairwise Relationships Between Numerical Variables
```

```python
sns.pairplot(animals.select_dtypes(include="number"))
plt.show()
```

```python
### Scatterplot Matrix Observations
##The strongest relationships are between the male and female measurements. Male and female size have a strong positive relationship, and male and female weight also have a strong positive relationship.
##The other numerical variables do not show relationships that are as clear. The barrier variables only have a few possible values, which is why their points appear in lines across the scatter plots.
```

```python
## Categorical vs. Categorical Variables
##To compare two categorical variables, animal class and conservation status are aggregated
```

```python
class_conservation = pd.crosstab(
    animals["class"],
    animals["conservation_status"]
)

class_conservation
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(class_conservation, annot=True)

plt.title("Animal Class vs. Conservation Status")
plt.xlabel("Conservation Status")
plt.ylabel("Animal Class")
plt.xticks(rotation=45)
plt.show()
```

```python
### Class and Conservation Status
##The heatmap compares animal class with conservation status. Mammalia is spread across several conservation categories, with the most animals being Least Concern and Endangered. Actinopterygii has the highest number of Least Concern animals. Some classes have much smaller counts because there are fewer animals from those classes in the dataset.
```

```python
sns.scatterplot(
    data=animals,
    x="size_male",
    y="size_female",
    hue="class"
)

plt.title("Male vs. Female Size by Animal Class")
plt.xlabel("Male Size")
plt.ylabel("Female Size")
plt.show()
```

```python
### Male and Female Size by Animal Class
##The graph shows a strong positive relationship between male and female size. As male size increases, female size usually increases as well. The colors show the different animal classes. Mammals make up many of the larger size measurements, while several of the other classes are mostly grouped around the smaller measurements.
```

```python
sns.boxplot(data=animals, x="class", y="size_male")

plt.title("Male Size by Animal Class")
plt.xlabel("Animal Class")
plt.ylabel("Male Size")
plt.xticks(rotation=45)
plt.show()
```

```python
### Male Size by Animal Class
#Male size varies between the different animal classes. Mammalia and Reptilia have a wider range of sizes, while classes such as Insecta and Amphibia are mostly made up of smaller animals. Some classes have very few observations, so their distributions are harder to compare.
```
